```python
# calculator with + - * / and divide-by-zero check

a=int(input("enter first number:"))
b=int(input("enter second number:"))
op=input("enter operator (+,-,*,/):")
if op== "+":
   print("result is",a+b)
elif op=="-":
   print("result is",a-b)
elif op=="*":
   print("result is",a*b)
elif op== "/":
    if b==0:
       print("cannot divide by zero")
    else:
       print("result is",a/b)
else:
   print("invalid operator")
```

```python
# count odd numbers in a list

nums=[3,7,9,4]
count = 0
for n in nums:
    if n%2==1:
       count=count+1
print("odd number",count)
```

```python
# find the smallest number in a list

nums=[4,17,9,25,6]
smallest=nums[0]
for n in nums:
    if n<smallest:
       smallest=n
print("smallest is",smallest)
```

```python
#function to check even or odd

def check(n):
    if n%2==0:
       print(n,"is even")
    else:
       print(n,"is odd")
check(8)
check(7)
```

```python
# function that returns the sum

def add(a,b):
    return(a+b)

result = add(5,3)
print("sum is",result)
print("double is", result*2)
```

```python
# function that returns the area, and compare two areas

def area(length,width):
    return(length*width)

result = area(4,5)
print("area is",result)
a = area(5,4)
b = area(3,9)
if a >= b:
   print("first is bigger")
else:
   print("second is bigger")
```
