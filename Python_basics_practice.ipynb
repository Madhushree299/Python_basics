```python
# calculator with + - * / and divide-by-zero check

a=int(input("enter first number:"))
b=int(input("enter second number:"))
op=input("enter operator (+,-,*,/):")
if op== "+":
   print("result is",a+b)
elif op=="-":
   print("result is",a-b)
elif op=="*":
   print("result is",a*b)
elif op== "/":
    if b==0:
       print("cannot divide by zero")
    else:
       print("result is",a/b)
else:
   print("invalid operator")
```

```python
# even or odd checker

a=int(input("enter number:"))
if a % 2 == 0:
   print("even")
else:
   print("odd")
```

```python
# sum and average of a list

marks=[10,20,30,40,50,60]
total=0
for m in marks:
   total=total+m
print("average is",total/len(marks))
```

```python
# count odd numbers in a list

nums=[3,7,9,4]
count = 0
for n in nums:
    if n%2==1:
       count=count+1
print("odd number",count)
```

```python
# find the smallest number in a list

nums=[4,17,9,25,6]
smallest=nums[0]
for n in nums:
    if n<smallest:
       smallest=n
print("smallest is",smallest)
```
